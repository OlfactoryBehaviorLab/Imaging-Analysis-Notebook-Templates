```python
from pathlib import Path

import numpy as np
import pandas as pd

from scipy.stats import pearsonr
```

```python
vgat_data = Path(r"R:\2_Inscopix\1_DTT\5_Combined\ID\VGAT_Comb\Analysis\Output\correlations.xlsx")
vgat = pd.read_excel(vgat_data, index_col=[0])
vglut_data = Path(r"R:\2_Inscopix\1_DTT\5_Combined\ID\VGLUT_Comb\Analysis\Output\correlations.xlsx")
vglut = pd.read_excel(vglut_data, index_col=[0])
ephys_data = Path(r"R:\5_Projects\6_DTT\E-Phys\Wilcoxon_p0.01-auROC_p0.02\Identity\odor_distance_selected.csv")
ephys = pd.read_csv(ephys_data, index_col=[0])
odor_distance = Path(r"R:\8_Odor_Space\chemical_distance_pearsons.xlsx")
odor = pd.read_excel(odor_distance, index_col=[0])
```

```python
ephysodor = pearsonr(odor, ephys, axis=None)
vgat_v_odor = pearsonr(odor, vgat, axis=None)
vglut_v_odor = pearsonr(odor, vglut, axis=None)
```

```python
print(f"ephys vs. distance: {ephysodor}\n"
      f"vgat vs. distance: {vgat_v_odor}\n"
      f"vglut vs. distance: {vglut_v_odor}")
```

```python
def matrix_no_reflection(input_matrix: pd.DataFrame):
    """"Remove reflection and zeros from the 2d matrix"""
    all_data = []

    for i, (name, data) in enumerate(input_matrix.iterrows()):
        trimmed_data = data.iloc[i+1:]
        all_data.append(trimmed_data.to_numpy())

    return np.hstack(all_data)
```

```python
flat_ephys = matrix_no_reflection(ephys)
flat_vgat = matrix_no_reflection(vgat)
flat_vglut = matrix_no_reflection(vglut)
flat_odor = matrix_no_reflection(odor)
```

```python
ephys_v_odor_flat = pearsonr(flat_odor, flat_ephys)
vgat_v_odor_flat = pearsonr(flat_odor, flat_vgat)
vglut_v_odor_flat = pearsonr(flat_odor, flat_vglut)
```

```python
print(f"ephys vs. distance: {ephys_v_odor_flat}\n"
      f"vgat vs. distance: {vgat_v_odor_flat}\n"
      f"vglut vs. distance: {vglut_v_odor_flat}")
```
