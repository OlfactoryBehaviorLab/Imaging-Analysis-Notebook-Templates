```python
%load_ext autoreload
%autoreload 2

import os
from pathlib import Path
os.environ['ISX'] = '0'

import pandas as pd
from dewan_calcium import plotting
import matplotlib.pyplot as plt
```

```python
concentration_data = Path(r"R:\5_Projects\6_DTT\E-Phys\Wilcoxon_p0.01-auROC_p0.02\Concentration\conc_heatmap_zscores_all.csv")
identity_data = Path(r"R:\5_Projects\6_DTT\E-Phys\Wilcoxon_p0.01-auROC_p0.02\Identity\heatmap_zscores_all.csv")


id_odor_order = ['But-AL', 'But-AMI', 'But-ATE', 'But-OL', 'But-ONE',
                 'Pent-AL', 'Pent-AMI', 'Pent-ATE', 'Pent-OL', 'Pent-ONE',
                 'Hex-AL', 'Hex-AMI', 'Hex-ATE', 'Hex-OL', 'Hex-ONE',
                 'Hept-AL', 'Hept-AMI', 'Hept-ATE', 'Hept-OL', "Hept-ONE",
                 "MO"] # Matches our ordering

conc_odor_order = ['AL-1', 'AL-10', 'AL-100', 'AL-1000',
                   'AMI-1', 'AMI-10', 'AMI-100', 'AMI-1000',
                   'ATE-1', 'ATE-10', 'ATE-100', 'ATE-1000',
                   'OL-1', 'OL-10', 'OL-100', 'OL-1000',
                   'ONE-1', 'ONE-10', 'ONE-100', 'ONE-1000',
                   'MO']

concentration_df = pd.read_csv(concentration_data, index_col=[0])
identity_df = pd.read_csv(identity_data, index_col=[0])

concentration_df.columns = [val.replace(chr(183), '-') for val in concentration_df.columns]

conc_means = concentration_df.loc[:, conc_odor_order].mean(axis=1)
conc_cell_order = conc_means.sort_values(ascending=False).index
id_means = identity_df.loc[:, id_odor_order].mean(axis=1)
id_cell_order = id_means.sort_values(ascending=False).index

sorted_conc_df = concentration_df.loc[conc_cell_order, conc_odor_order]
sorted_id_df = identity_df.loc[id_cell_order, id_odor_order]

plotting.plot_all_avg_dff(sorted_conc_df, concentration_data.parent)
plotting.plot_all_avg_dff(sorted_id_df, identity_data.parent)
```

```python
polar_plot_data = Path(r"R:\5_Projects\6_DTT\E-Phys\Wilcoxon_p0.01-auROC_p0.02\Sniffing\Identity+Concentration\PreferredAnglevsR_prevsPost.xlsx")

polar_plot_df = pd.read_excel(polar_plot_data)

pre_theta = polar_plot_df.loc[:, ["pref_pre_deg"]]
pre_r = polar_plot_df.loc[:, ["R_pre"]]
post_theta = polar_plot_df.loc[:, ["pref_post_deg"]]
post_r = polar_plot_df.loc[:, ["R_post"]]

average_pre_theta = pre_theta.mean().to_numpy()[0]
average_pre_r = pre_r.mean().to_numpy()[0]
average_post_theta = post_theta.mean().to_numpy()[0]
average_post_r = post_r.mean().to_numpy()[0]
```

```python
import numpy as np
fig, ax = plt.subplots(subplot_kw={'projection': 'polar'})
pre = ax.scatter(pre_theta, pre_r, s=0.8)
post = ax.scatter(post_theta, post_r, s=0.8)
ax.set_rticks([0.1, 0.3, 0.5, 0.7])  # Fewer radial ticks
ax.set_theta_zero_location("N")
ax.set_theta_direction(-1)
# ax.arrow(0, 0, average_pre_theta,1, width=0.2)


ax.annotate("", xy=(average_pre_theta, .5), xytext=(0, 0),
            arrowprops=dict(arrowstyle="->", lw=2, color=pre.get_edgecolor()))

ax.annotate("", xy=(average_post_theta, .5), xytext=(0, 0),
            arrowprops=dict(arrowstyle="->", lw=2, color=post.get_edgecolor()))
plt.show()
```
